# Creating a Tensor

```python
import tensorflow as tf
```

```python
scaler=tf.constant(42)
print(scaler)
print("Scaler shape:",scaler.shape)
print(type(scaler))
```

```python
# create a Tensor using a vactor
vector=tf.constant([1.0,2.0,3.0])
print(vector)
print("Vector shape:",vector.shape)
print(type(vector))
```

# Creating a Tensor Using a matrix

```python
vector=tf.constant([1.0,2.0,3.0])
print(vector)
print("Vector shape:",vector.shape)
print(type(vector))
```

```python
# create a Tensor Using a Matrix
matrix=tf.constant([[1,2],[3,4]])
print(matrix)
print("Matrix shape:",matrix.shape)
```

# 3D Tensor

```python
tensor_3d=tf.constant([[[1],[2]],[[3],[4]]])
print(tensor_3d)
print("3D Tensor shape:",tensor_3d.shape)
print(type(tensor_3d))
```

```python
# creating a Mutable tensor
mutable_tensor=tf.Variable([[1.0,2.0]])
print(mutable_tensor)
print(mutable_tensor.shape)
print(type(mutable_tensor))
```

```python
# Tensor with Zeros
z=tf.zeros([2,3])
print(z)
print(z.shape)
print(type(z))
```

```python
# Tensor with ones
one=tf.ones([2,3])
print(one)
print(one.shape)
print(type(one))
```

```python
# Tensor with random values
r=tf.random.normal([2,2])
print(r)
print(r.shape)
print(type(r))
```

# Tensor Operations
